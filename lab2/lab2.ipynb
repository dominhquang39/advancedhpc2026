```python
import numba
from numba import cuda

cuda.detect()
```

```python
device = cuda.select_device(0)

print(f"Device id: {device.id}")
print(f"Device name: {device.name}")
```

Core info

```python
multiprocessor_count = device.MULTIPROCESSOR_COUNT
cores_per_SM = 64 # Look up CUDA cores per SM from dict based on Compute Capability (64 for Compute Capability 7.5)
total_cores = multiprocessor_count * cores_per_SM

print(f"Multiprocessor count: {multiprocessor_count}")
print(f"Total cores: {total_cores}")
```

Memory Info

```python
mem_info = cuda.current_context().get_memory_info()
print(f"Total memory: {mem_info[1]}")
```
